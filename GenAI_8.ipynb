```python
# 1. Install required dependencies
!pip install -q diffusers transformers accelerate

import torch
from diffusers import StableDiffusionPipeline

# 2. Load the model pipeline
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
)

# 3. Move pipeline to GPU
pipe = pipe.to("cuda")

# 4. Generate the image
prompt = "A futuristic city skyline at sunset, digital art, highly detailed"
image = pipe(
    prompt,
    num_inference_steps=30,
    guidance_scale=7.5
).images[0]

# 5. Save and display
image.save("generated_city.png")
print("Image generated and saved as generated_city.png")

# Display inside Colab notebook
display(image)
```
