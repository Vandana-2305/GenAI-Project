```python
# Install required dependencies
!pip install -q gTTS diffusers transformers

import torch
from diffusers import StableDiffusionPipeline
from gtts import gTTS
from transformers import pipeline


def main():
    # Setup hardware device
    device = "cuda" if torch.cuda.is_available() else "cpu"
    torch_dtype = torch.float16 if device == "cuda" else torch.float32

    topic = "The benefits of renewable energy"
    print(f"--- Processing Topic: '{topic}' ---\n")

    # 1. Text Generation
    print("[1/3] Generating text...")
    text_generator = pipeline(
        "text-generation", model="google/flan-t5-base"
    )
    text_prompt = f"Write a short, engaging paragraph about: {topic}"
    text_output = text_generator(text_prompt, max_new_tokens=80)
    generated_text = text_output[0]["generated_text"]

    print(f"Generated Text:\n{generated_text}\n")

    # 2. Image Generation
    print("[2/3] Generating image...")
    image_prompt = f"An illustration representing {topic}, digital art"
    sd_pipe = StableDiffusionPipeline.from_pretrained(
        "runwayml/stable-diffusion-v1-5", torch_dtype=torch_dtype
    ).to(device)

    image = sd_pipe(image_prompt, num_inference_steps=25).images[0]
    image.save("content_image.png")
    print("Saved image: content_image.png\n")

    # 3. Audio Generation (Text-to-Speech)
    print("[3/3] Generating audio...")
    tts = gTTS(text=generated_text, lang="en")
    tts.save("content_audio.mp3")
    print("Saved audio: content_audio.mp3\n")

    print("--- Execution Complete ---")


if __name__ == "__main__":
    main()
```
