```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM, AutoModelForQuestionAnswering
import torch

# ---------- Text Summarization ----------
# Load tokenizer and model directly
tokenizer = AutoTokenizer.from_pretrained("facebook/bart-large-cnn")
model = AutoModelForSeq2SeqLM.from_pretrained("facebook/bart-large-cnn")

# Prepare the input for the model
article = """Generative AI refers to a class of artificial intelligence models capable of
producing new content such as text, images, audio, and video. Large Language Models (LLMs)
such as GPT and LLaMA are trained on massive text corpora and can perform a wide range of
natural language tasks including translation, summarization, and question answering. These
models are increasingly being deployed in industry applications ranging from customer support
to software development, transforming how humans interact with machines."""
input_ids = tokenizer(article, return_tensors="pt").input_ids

# Generate summary directly using the model
summary_ids = model.generate(
    input_ids,
    max_length=45,
    min_length=20,
    num_beams=4,  # Using beam search for better summary quality
    do_sample=False,
    early_stopping=True
)
summary_text = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
print("Summary:\n", summary_text)

# ---------- Question Answering ----------
# Reverting to direct model usage for text-based Question Answering
qa_tokenizer = AutoTokenizer.from_pretrained("distilbert-base-cased-distilled-squad")
qa_model = AutoModelForQuestionAnswering.from_pretrained("distilbert-base-cased-distilled-squad")

context = article
question = "What are Large Language Models trained on?"

# Tokenize input
inputs = qa_tokenizer(question, context, return_tensors="pt")

# Get model outputs
outputs = qa_model(**inputs)

# Get the most likely start and end token of the answer
answer_start_index = outputs.start_logits.argmax()
answer_end_index = outputs.end_logits.argmax()

# Decode the answer
predict_answer_tokens = inputs.input_ids[0, answer_start_index : answer_end_index + 1]
answer = qa_tokenizer.decode(predict_answer_tokens, skip_special_tokens=True)

# For confidence (simplified, as raw logits are not probabilities directly)
# A common way to get a proxy for confidence is to look at the logit scores
# Higher scores mean more confident prediction. This is not a true probability.
# For simplicity, we'll just report the raw score for now.
start_score = torch.max(outputs.start_logits).item()
end_score = torch.max(outputs.end_logits).item()
confidence_score = (start_score + end_score) / 2 # Simple average of max logits

print("\nQuestion:", question)
print("Answer:", answer, "| Confidence (logit avg):", round(confidence_score, 3))
```
